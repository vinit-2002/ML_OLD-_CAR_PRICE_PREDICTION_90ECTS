```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer

# Load dataset
df = pd.read_csv('/kaggle/input/datasets/nehalbirla/vehicle-dataset-from-cardekho/car data.csv')

# Clean column names
df.columns = df.columns.str.strip().str.lower()

# Display available columns
print("Available columns:", df.columns.tolist())

# Remove duplicate rows
df = df.drop_duplicates().copy()

# Extract brand if name exists
if 'name' in df.columns:
    df['brand'] = df['name'].astype(str).str.split().str[0]

# Convert available numeric-text columns
for col in ['mileage', 'engine', 'max_power']:
    if col in df.columns:
        df[col] = pd.to_numeric(
            df[col].astype(str).str.extract(r'(\d+\.?\d*)')[0],
            errors='coerce'
        )

# Create car age if year exists
if 'year' in df.columns:
    df['car_age'] = 2026 - df['year']

# Define target
target = 'selling_price'

if target not in df.columns:
    raise ValueError("The dataset does not contain a selling_price column.")

# Select available features
features = [
    col for col in [
        'brand', 'year', 'km_driven', 'fuel',
        'seller_type', 'transmission', 'owner',
        'mileage', 'engine', 'max_power', 'seats', 'car_age'
    ]
    if col in df.columns
]

X = df[features]
y = df[target]

# Identify column types
numerical_features = X.select_dtypes(include=np.number).columns.tolist()
categorical_features = X.select_dtypes(include='object').columns.tolist()

# Preprocessing
numerical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer([
    ('num', numerical_transformer, numerical_features),
    ('cat', categorical_transformer, categorical_features)
])

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print("Setup completed successfully!")
print("Features:", features)
print("Training shape:", X_train.shape)
print("Testing shape:", X_test.shape)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

results = []

def evaluate_model(model, model_name):
    y_pred = model.predict(X_test)

    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)

    print("\nMODEL:", model_name)
    print("MAE:", round(mae, 2))
    print("MSE:", round(mse, 2))
    print("RMSE:", round(rmse, 2))
    print("R2 Score:", round(r2, 4))
    print("R2 Performance:", round(r2 * 100, 2), "%")

    results.append({
        'Model': model_name,
        'MAE': mae,
        'MSE': mse,
        'RMSE': rmse,
        'R2 Score': r2,
        'Performance (%)': r2 * 100
    })
```

```python
from sklearn.linear_model import LinearRegression

lr_model = Pipeline([
    ('preprocessor', preprocessor),
    ('model', LinearRegression())
])

lr_model.fit(X_train, y_train)

evaluate_model(lr_model, "Linear Regression")
```

```python
from sklearn.tree import DecisionTreeRegressor

dt_model = Pipeline([
    ('preprocessor', preprocessor),
    ('model', DecisionTreeRegressor(
        max_depth=8,
        min_samples_split=5,
        min_samples_leaf=2,
        random_state=42
    ))
])

dt_model.fit(X_train, y_train)

evaluate_model(dt_model, "Decision Tree Regression")
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = Pipeline([
    ('preprocessor', preprocessor),
    ('model', RandomForestRegressor(
        n_estimators=100,
        max_depth=12,
        min_samples_split=5,
        min_samples_leaf=2,
        random_state=42,
        n_jobs=-1
    ))
])

rf_model.fit(X_train, y_train)

evaluate_model(rf_model, "Random Forest Regression")
```

```python
from sklearn.ensemble import AdaBoostRegressor

ada_model = Pipeline([
    ('preprocessor', preprocessor),
    ('model', AdaBoostRegressor(
        n_estimators=100,
        learning_rate=0.05,
        loss='linear',
        random_state=42
    ))
])

ada_model.fit(X_train, y_train)

evaluate_model(ada_model, "AdaBoost Regression")
```

```python
comparison_df = pd.DataFrame(results)

comparison_df = comparison_df.sort_values(
    by='RMSE',
    ascending=True
)

display(comparison_df.round(2))

best_model_name = comparison_df.iloc[0]['Model']

print("Best Performing Model:", best_model_name)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))

sns.barplot(
    data=comparison_df,
    x='Model',
    y='R2 Score'
)

plt.title('Comparison of Machine Learning Models')
plt.xlabel('Machine Learning Models')
plt.ylabel('R2 Score')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
from sklearn.ensemble import GradientBoostingRegressor

gb_model = Pipeline([
    ('preprocessor', preprocessor),
    ('model', GradientBoostingRegressor(
        n_estimators=100,
        learning_rate=0.05,
        max_depth=3,
        random_state=42
    ))
])

gb_model.fit(X_train, y_train)

evaluate_model(gb_model, "Gradient Boosting Regression")
```
